# Customer Analytics, Repeat Purchase Behavior & Geographic Demand
**Objective**: Differentiate `customer_id` from `customer_unique_id`, measure repeat buyer rates, and analyze geographic customer concentration.

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

proc_dir = r'../data/processed'
rfm = pd.read_csv(os.path.join(proc_dir, 'customer_rfm.csv'))

repeat_count = (rfm['frequency'] > 1).sum()
total_cust = len(rfm)
print(f'Total Customers: {total_cust:,}')
print(f'Repeat Buyers: {repeat_count:,} ({repeat_count/total_cust*100:.2f}%)')

state_dist = rfm['customer_state'].value_counts().head(10)
plt.figure(figsize=(10, 4))
sns.barplot(x=state_dist.index, y=state_dist.values, palette='viridis')
plt.title('Top 10 Brazilian States by Customer Base')
plt.ylabel('Unique Customer Count')
plt.xlabel('State')
plt.tight_layout()
plt.show()
```
